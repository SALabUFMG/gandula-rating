```python
from statsbombpy import sb
import pandas as pd
from pathlib import Path
import logging

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

class StatsBombLoader:
    """Class to load and process StatsBomb data using statsbombpy."""
    
    def __init__(self, data_dir: str = "data/statsbomb"):
        """
        Initialize the StatsBomb loader.
        
        Args:
            data_dir (str): Directory to store the downloaded data
        """
        self.data_dir = Path(data_dir)
        self.data_dir.mkdir(parents=True, exist_ok=True)
        
    def get_competitions(self) -> pd.DataFrame:
        """
        Get all available competitions.
        
        Returns:
            pd.DataFrame: DataFrame containing competitions information
        """
        try:
            competitions = pd.DataFrame(sb.competitions())
            competitions.to_csv(self.data_dir / "competitions.csv", index=False)
            logger.info(f"Successfully downloaded competitions data")
            return competitions
        except Exception as e:
            logger.error(f"Error downloading competitions data: {e}")
            raise
    
    def get_matches(self, competition_id: int, season_id: int) -> pd.DataFrame:
        """
        Get matches data for a specific competition and season.
        
        Args:
            competition_id (int): ID of the competition
            season_id (int): ID of the season
            
        Returns:
            pd.DataFrame: DataFrame containing matches information
        """
        try:
            matches = pd.DataFrame(sb.matches(competition_id=competition_id, season_id=season_id))
            matches.to_csv(self.data_dir / f"matches_{competition_id}_{season_id}.csv", index=False)
            logger.info(f"Successfully downloaded matches data for competition {competition_id}, season {season_id}")
            return matches
        except Exception as e:
            logger.error(f"Error downloading matches data: {e}")
            raise
    
    def get_events(self, match_id: int) -> pd.DataFrame:
        """
        Get events data for a specific match.
        
        Args:
            match_id (int): ID of the match
            
        Returns:
            pd.DataFrame: DataFrame containing events information
        """
        try:
            events = pd.DataFrame(sb.events(match_id=match_id))
            events.to_csv(self.data_dir / f"events_{match_id}.csv", index=False)
            logger.info(f"Successfully downloaded events data for match {match_id}")
            return events
        except Exception as e:
            logger.error(f"Error downloading events data: {e}")
            raise
    
    def get_lineups(self, match_id: int) -> pd.DataFrame:
        """
        Get lineups data for a specific match.
        
        Args:
            match_id (int): ID of the match
            
        Returns:
            pd.DataFrame: DataFrame containing lineups information
        """
        try:
            lineups = pd.DataFrame(sb.lineups(match_id=match_id))
            lineups.to_csv(self.data_dir / f"lineups_{match_id}.csv", index=False)
            logger.info(f"Successfully downloaded lineups data for match {match_id}")
            return lineups
        except Exception as e:
            logger.error(f"Error downloading lineups data: {e}")
            raise


"""Example usage of the StatsBombLoader class."""
# Initialize the loader
loader = StatsBombLoader()

# Get competitions data
competitions = loader.get_competitions()
print("\nAvailable competitions:")
print(competitions[['competition_id', 'competition_name', 'country_name']])
```

## VAEP Features

```python
from socceraction.vaep import features as ft
```

```python
def construct_features(actions, keep_result=True):
    actions['nickname'] = actions['nickname'].fillna('')
    actions['original_event_id'] = actions['original_event_id'].fillna('')

    xfns = [
        #ft.actiontype,
        ft.actiontype_onehot,
        #ft.bodypart,
        ft.bodypart_onehot,
        #ft.result,
        ft.goalscore,
        ft.startlocation,
        ft.endlocation,
        ft.movement,
        ft.space_delta,
        ft.startpolar,
        ft.endpolar,
        ft.team,
        ft.time,
        ft.time_delta
    ]

    if keep_result:
        xfns.append(ft.result_onehot)

    match_states = ft.gamestates(actions=actions)
    match_feats = pd.concat([fn(match_states) for fn in xfns], axis=1)

    # features = pd.concat(features).reset_index(drop=True)
    features = pd.concat([actions[['game_id', 'action_id']], match_feats], axis=1)

    return features
```

```python
features_actual = construct_features(match_actions)
features_expected = construct_features(match_actions, keep_result=False)
```

## VAEP Labels Goal

```python
from socceraction.vaep import labels as lb
```

```python
def construct_labels(actions):
  yfns = [lb.scores, lb.concedes]

  labels = pd.concat([fn(actions=actions) for fn in yfns], axis=1)

  labels = pd.concat([actions[['game_id', 'action_id']], labels], axis=1)

  return labels
```

```python
labels = construct_labels(match_actions)
```
